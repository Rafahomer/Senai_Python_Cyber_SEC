## Inicio do Cadastro ativo de rede


```python
print ("=" * 50)
print ("    NETsENTINEL - Cadastro de ativos de rede  ")
print ("=" * 50)

# Inserindo as strings
nome_host = input("Digite o nome do Host (hostname): ")
ip_host = input("Digite o IP do Host (ex: 192.168.1.10): ")
mac_host = input("Digite o MAC do Host (ex: XX:XX:XX:XX:XX:XX): ")
sistema_operacional = input("Digite o sistema operacional do Host (linux ou windows): ")

# Inserindo os inteiros
porta_critica = int(input("Digite a principal porta aberta a monitorar (ex: 22, 80, 5060): "))
nivel_de_criticidade = int(input("Informe o nível de criticidade do ativo (1 - baixo a 5 - crítico): "))

# Inserindo os Floats (valores decimais)
tempo_de_resposta_ping = float(input("Digite o tempo médio de ping em ms: "))
score_cvss = float(input("Pontuação média de vulnerabilidade CVSS: "))
# Desafio 1: Consumo de banda usada pelo ativo em Mbps
consumo_banda = float(input("Digite o consumo de banda atual em Mbps: "))

# Valores booleanos (True ou False)
firewall_ativo_input = input("O Firewall está ativo? (s/n): ").strip().lower()
firewall_ativo = (firewall_ativo_input == "s")

status_ativo_input = input("O Host está acessível? (s/n): ").strip().lower()
status_ativo = (status_ativo_input == "s")
```

## Geração Do relatorio


```python
print ("=" * 50)
print ("    Relatório do Host Cadastrado  ")
print ("=" * 50)

# Gerar o Relatório
print(f'Hostname: {nome_host} (tipo: {type(nome_host).__name__})')
print(f'Endereço IP: {ip_host} (tipo: {type(ip_host).__name__})')
print(f'Endereço MAC: {mac_host} (tipo: {type(mac_host).__name__})')
print(f'Sistema Operacional: {sistema_operacional} (tipo: {type(sistema_operacional).__name__})')
print(f'Porta Alvo: {porta_critica} (tipo: {type(porta_critica).__name__})')
print(f'Porta Alvo Nível de Criticidade: {nivel_de_criticidade}/5')
# Desafio 3: Formatar casas decimais para o tempo de Ping (ex: 2 casas decimais com .2f)
print(f'Latência (ping): {tempo_de_resposta_ping:.2f} ms (tipo: {type(tempo_de_resposta_ping).__name__})')
print(f'Score Risco (CVSS): {score_cvss}/10.0')
# Desafio 1: Exibindo o consumo de banda
print(f'Consumo de Banda: {consumo_banda:.2f} Mbps (tipo: {type(consumo_banda).__name__})')
print(f'Firewall Ativo: {firewall_ativo} (tipo: {type(firewall_ativo).__name__})')
print(f'Host Acessível: {status_ativo} (tipo: {type(status_ativo).__name__})')
```

##PRIMEIRA REGRA DE SEGURANÇA


```python
# Verificação de score
print("=" * 50)

# Desafio 2: Verificação se a porta inserida é uma porta padrão de acesso remoto inseguro
# Exemplo: 21 (FTP), 23 (Telnet), 3389 (RDP sem criptografia forte por padrão em versões antigas)
portas_inseguras = [21, 23, 3389]

if porta_critica in portas_inseguras:
    print(f"⚠️ ALERTA: A porta {porta_critica} monitorada é uma porta conhecida de acesso remoto inseguro!")

alerta_sec = (score_cvss >= 7.0) or (not firewall_ativo)

print (f"Alerta de Segurança Requerido: {alerta_sec}")
if alerta_sec:
    print("ATENÇÃO: Este host possui alto risco de segurança ou o firewall está desativado!")
else:
    print("STATUS: Host dentro dos parâmetros de conformidade")

print("=" * 50)
```

##Calculadora de Criticidade


```python
nivel_criticidade = (score_cvss >= 7) or (not firewall_ativo)
pontuacao = 0

if score_cvss >= 8:
  pontuacao += 5

if not firewall_ativo:
  pontuacao += 3

if porta_critica == 22: #SSH
  pontuacao += 2

if porta_critica == 3389: #RDP
  pontuacao += 3

if nivel_de_criticidade >= 4:
  pontuacao += 4

print('======== RESULTADO ========')
print(f'Hostname: {nome_host}')
print(f'Pontuação: {pontuacao}')
```
