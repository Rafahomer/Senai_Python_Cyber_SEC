```python
# ========================================== #
#         NETSENTINEL - CÓDIGO FONTE         #
# ========================================== #

# --- 1. CADASTRO DE ATIVOS ---
print("=" * 50)
print("    NETSENTINEL - Cadastro de ativos de rede  ")
print("=" * 50)

nome_host = input("Digite o nome do Host (hostname): ")
ip_host = input("Digite o IP do Host (ex: 192.168.1.10): ")
mac_host = input("Digite o MAC do Host (ex: XX:XX:XX:XX:XX:XX): ")
sistema_operacional = input("Digite o sistema operacional do Host (linux ou windows): ")
porta_critica = int(input("Digite a principal porta aberta a monitorar (ex: 22, 80, 5060, 23): "))
nivel_de_criticidade = int(input("Informe o nível de criticidade do ativo (1 - baixo a 5 - crítico): "))
tempo_de_resposta_ping = float(input("Digite o tempo médio de ping em ms: "))
score_cvss = float(input("Pontuação média de vulnerabilidade CVSS: "))
consumo_banda = float(input("Digite o consumo de banda atual em Mbps: "))
firewall_ativo_input = input("O Firewall está ativo? (s/n): ").strip().lower()
firewall_ativo = (firewall_ativo_input == "s")
status_ativo_input = input("O Host está acessível? (s/n): ").strip().lower()
status_ativo = (status_ativo_input == "s")

# --- 2. RELATÓRIO DO HOST ---
print("\n" + "=" * 50)
print("    Relatório do Host Cadastrado  ")
print("=" * 50)

print(f'Hostname: {nome_host} (tipo: {type(nome_host).__name__})')
print(f'Endereço IP: {ip_host} (tipo: {type(ip_host).__name__})')
print(f'Endereço MAC: {mac_host} (tipo: {type(mac_host).__name__})')
print(f'Sistema Operacional: {sistema_operacional} (tipo: {type(sistema_operacional).__name__})')
print(f'Porta Alvo: {porta_critica} (tipo: {type(porta_critica).__name__})')
print(f'Porta Alvo Nível de Criticidade: {nivel_de_criticidade}/5')
print(f'Latência (ping): {tempo_de_resposta_ping:.2f} ms (tipo: {type(tempo_de_resposta_ping).__name__})')
print(f'Score Risco (CVSS): {score_cvss}/10.0')
print(f'Consumo de Banda: {consumo_banda:.2f} Mbps (tipo: {type(consumo_banda).__name__})')
print(f'Firewall Ativo: {firewall_ativo} (tipo: {type(firewall_ativo).__name__})')
print(f'Host Acessível: {status_ativo} (tipo: {type(status_ativo).__name__})')

# --- 3. AVALIAÇÃO DE VULNERABILIDADES ISOLADAS (Desafios 1, 4 e 5) ---
print("\n" + "=" * 50)
print("    ANÁLISE DE ALERTAS ESPECÍFICOS")
print("=" * 50)

# Desafio 1 e 2: Detectando Telnet
if porta_critica == 23:
    print("ALERTA: Serviço Telnet (Porta 23) detectado! Exposição de credenciais em texto claro.")
else:
    print("Porta principal fora do espectro crítico do Telnet.")

# Desafio 4: Avaliando CVSS
if score_cvss >= 7.0:
    print("ALERTA: Ativo possui vulnerabilidade alta (CVSS >= 7.0)!")
else:
    print("Nível de vulnerabilidade CVSS dentro do limite de segurança aceitável (CVSS < 7.0).")

# Desafio 5: Firewall Desativado
if not firewall_ativo:
    print("ATENÇÃO: O ativo encontra-se sem o firewall ativado.")
else:
    print("Segurança local ativa: Firewall do host está ativado.")

# Desafio 6: Alerta Crítico Combinado
if score_cvss >= 8.0 and not firewall_ativo:
    print("ALERTA CRÍTICO: Vulnerabilidade grave detectada E firewall desativado!")
else:
    print("Não há ocorrência simultânea de alta vulnerabilidade e firewall desativado.")

# --- 4. CYBER ALERT ENGINE - MOTOR DE ALERTAS (Desafios 3 e Integrador) ---
print("\n" + "=" * 50)
print("    CYBER ALERT ENGINE - RESULTADO FINAL")
print("=" * 50)

# Lógica principal de classificação integrando todos os fatores
if porta_critica == 23 or score_cvss >= 9.0 or (score_cvss >= 8.0 and not firewall_ativo):
    risco = 'RISCO CRÍTICO'
    acao = 'Investigar imediatamente. Isolar o host da rede.'
elif porta_critica == 3389 or score_cvss >= 7.0 or not firewall_ativo:
    risco = 'RISCO ALTO'
    acao = 'Aplicar correções de segurança (patches) e verificar regras de firewall.'
elif porta_critica == 22:
    risco = 'RISCO MÉDIO'
    acao = 'Monitorizar logs de acesso SSH em busca de brute-force.'
else:
    risco = 'RISCO BAIXO'
    acao = 'Manter monitorização padrão de rotina.'

# Saída de Dados do Motor de Alertas
print(f'Hostname: {nome_host}')
print(f'Porta Analisada: {porta_critica}')
print(f'Score CVSS: {score_cvss}')
print(f'Status Firewall: {"Ativo" if firewall_ativo else "Inativo"}')
print('\nClassificação Final:')
print(f'>> {risco} <<')
print('\nAção Recomendada para o Analista:')
print(f'>> {acao}')
print("=" * 50)
```
