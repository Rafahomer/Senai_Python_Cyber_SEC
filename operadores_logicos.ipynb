##Operadores Logicos



### Explicação dos Operadores Lógicos

Os operadores lógicos são usados para combinar expressões condicionais em Python. Eles avaliam as expressões e retornam um valor booleano (`True` ou `False`).

1. **`and` (E)**: Retorna `True` apenas se **todas** as condições forem verdadeiras.
   * Exemplo: `x > 2 and x < 5` (Verdadeiro se `x` for 3 ou 4).

2. **`or` (OU)**: Retorna `True` se **pelo menos uma** das condições for verdadeira.
   * Exemplo: `y < 6 or y < 5` (Retorna `True` se qualquer uma das duas condições for atendida).

3. **`not` (NÃO)**: Inverte o valor lógico da expressão. Se o resultado for `True`, ele se torna `False`, e vice-versa.
   * Exemplo: `not(x > 3)`.

```python
x = 3
y = 6
print (x>3)
print (y<6)
print ("="*50)
print (x>=3)
print (y<=6)
print ("="*50)
print (x>3 and x<2)
print (y<6 or y<5)
```

```python
# Exemplo de uso do operador 'not'
tem_sol = True

# O operador 'not' inverte o valor booleano
print("Está sol?", tem_sol)
print("Está chovendo (não tem sol)?", not tem_sol)

print("="*50)

# Exemplo com comparação
idade = 15
maior_de_idade = idade >= 18
print("É maior de idade?", maior_de_idade)
print("É menor de idade (não é maior de idade)?", not maior_de_idade)
```
