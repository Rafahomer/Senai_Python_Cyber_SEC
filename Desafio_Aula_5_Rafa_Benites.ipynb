```python
# ========================================== #
#         NETSENTINEL - CÓDIGO FONTE         #
# ========================================== #

print("=" * 50)
print("    NETSENTINEL - Painel do SOC (Monitoramento)  ")
print("=" * 50)

# Lista para armazenar múltiplos ativos (Inventário)
inventario_ativos = []
cadastrando = True

# --- 1. ESTRUTURA WHILE: CADASTRO CONTÍNUO DE ATIVOS ---
while cadastrando:
    print("\n--- Novo Cadastro de Ativo ---")
    nome_host = input("Digite o nome do Host (hostname): ")
    ip_host = input("Digite o IP do Host (ex: 192.168.1.10): ")
    mac_host = input("Digite o MAC do Host (ex: XX:XX:XX:XX:XX:XX): ")
    sistema_operacional = input("Digite o SO do Host (linux ou windows): ")
    porta_critica = int(input("Digite a principal porta aberta (ex: 22, 80, 5060, 23): "))
    nivel_de_criticidade = int(input("Nível de criticidade (1 - baixo a 5 - crítico): "))
    tempo_de_resposta_ping = float(input("Tempo médio de ping em ms: "))
    score_cvss = float(input("Pontuação média de vulnerabilidade CVSS: "))
    consumo_banda = float(input("Consumo de banda atual em Mbps: "))

    firewall_ativo = input("O Firewall está ativo? (s/n): ").strip().lower() == "s"
    status_ativo = input("O Host está acessível? (s/n): ").strip().lower() == "s"

    # Salva o ativo atual como um dicionário dentro da lista
    ativo = {
        "hostname": nome_host,
        "ip": ip_host,
        "mac": mac_host,
        "so": sistema_operacional,
        "porta": porta_critica,
        "criticidade": nivel_de_criticidade,
        "ping": tempo_de_resposta_ping,
        "cvss": score_cvss,
        "banda": consumo_banda,
        "firewall": firewall_ativo,
        "online": status_ativo
    }

    inventario_ativos.append(ativo)

    # Condição de parada do While
    continuar = input("\nDeseja cadastrar outro ativo para monitoramento? (s/n): ").strip().lower()
    if continuar != 's':
        cadastrando = False

# --- 2. ESTRUTURA FOR: VARREDURA E MOTOR DE ALERTAS (SOC) ---
print("\n" + "=" * 50)
print("    INICIANDO VARREDURA DE SEGURANÇA (CYBER ENGINE)")
print("=" * 50)

# Loop FOR para analisar todos os ativos cadastrados automaticamente
for ativo in inventario_ativos:
    print(f"\n[{ativo['hostname']}] - Analisando IP: {ativo['ip']} ...")

    # Verifica status operacional
    if not ativo['online']:
        print(f" ❌ INCIDENTE: O ativo {ativo['hostname']} está OFFLINE. Varredura suspensa para este host.")
        print("-" * 50)
        continue # Pula para o próximo ativo se este estiver offline

    print(f" ✅ Status: ONLINE | Ping: {ativo['ping']}ms | Banda: {ativo['banda']} Mbps")

    # --- AVALIAÇÃO DE VULNERABILIDADES ISOLADAS ---
    if ativo['porta'] == 23:
        print(" ⚠️ ALERTA ESPECÍFICO: Serviço Telnet (Porta 23) detectado!")

    if ativo['cvss'] >= 7.0:
        print(f" ⚠️ ALERTA ESPECÍFICO: Vulnerabilidade alta (CVSS {ativo['cvss']})")

    if not ativo['firewall']:
        print(" ⚠️ ALERTA ESPECÍFICO: Firewall local desativado.")

    if ativo['cvss'] >= 8.0 and not ativo['firewall']:
        print(" 🚨 ALERTA CRÍTICO COMBINADO: Alta vulnerabilidade + Firewall inativo!")

    # --- CYBER ALERT ENGINE (Classificação Final) ---
    if ativo['porta'] == 23 or ativo['cvss'] >= 9.0 or (ativo['cvss'] >= 8.0 and not ativo['firewall']):
        risco = 'RISCO CRÍTICO'
        acao = 'Isolar o host da rede imediatamente. Risco iminente de invasão.'
    elif ativo['porta'] == 3389 or ativo['cvss'] >= 7.0 or not ativo['firewall']:
        risco = 'RISCO ALTO'
        acao = 'Aplicar patches de segurança e revisar regras de firewall.'
    elif ativo['porta'] == 22:
        risco = 'RISCO MÉDIO'
        acao = 'Monitorizar logs do OpenSSH em busca de ataques brute-force.'
    else:
        risco = 'RISCO BAIXO'
        acao = 'Manter monitorização padrão.'

    # Saída do Motor de Alertas
    print(f" 📊 Classificação Final: >> {risco} <<")
    print(f" 🛡️ Ação Recomendada: {acao}")
    print("-" * 50)

print("\nVARREDURA CONCLUÍDA. Todos os ativos foram analisados.")
print("=" * 50)
```
